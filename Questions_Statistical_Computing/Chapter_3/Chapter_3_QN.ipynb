```python
n = 100
df=pd.DataFrame()
np.random.seed(1)
df['X'] = np.random.randn(n)
df['Y'] = 2*df['X']+ np.random.randn(n)
```

```python
model_1=sm.OLS(df['Y'],df['X']).fit()
print(model_1.summary())
```

$$ {t} =  \frac{\hat {B}}{SE(\hat {B})} $$

$$ =\frac{\frac{\sum_{i=1}^{n}\left ( x_{i} y_{i} \right )}{\sum_{j=1}^{n} x_{j}^{2}}}{\sqrt {\frac{{\sum_{i'=1}^{n}\left (  y_{i'}-x_{i'}*\hat {B} \right )^2)}}{(n-1) \times \sum_{j'=1}^{n} x_{j}^{2}}} } $$

$$ =\frac{\sqrt{(n-1)} \times \sum_{i=1}^{n}\left ( x_{i} y_{i} \right )}{\sqrt{\sum_{j=1}^{n} x_{j}^{2} \times \sum_{i'=1}^{n}\left (  y_{i'}-x_{i'}*\hat {B} \right )^2)}}  $$

$$ \sum_{j=1}^{n} x_{j}^{2} \times \sum_{i'=1}^{n}\left (  y_{i'}-x_{i'}*\hat {B} \right )^2)=\sum_{j=1}^{n} x_{j}^{2} \times \sum_{i'=1}^{n}\left (  y_{i'}^2-2y_{i'}*x_{i'}*\hat {B}+(x_{i'}*\hat {B})^2 \right )) $$
$$ =\sum_{j=1}^{n} x_{j}^{2} \times \sum_{i'=1}^{n}\left (  y_{i'}^2-2y_{i'}*x_{i'}*\frac{\sum_{i'=1}^{n}\left ( x_{i'} y_{i'} \right )}{\sum_{j=1}^{n} x_{j}^{2}}+(x_{i'})^2*(\frac{\sum_{i'=1}^{n}\left ( x_{i'} y_{i'} \right )}{\sum_{j=1}^{n} x_{j}^{2}})^2 \right )) $$
$$=\sum_{j=1}^{n} x_{j}^{2} \times \sum_{j=1}^{n} y_{j}^{2} - \sum_{i=1}^{n}\left ( x_{i} y_{i} \right )^2 $$

$$ {t} =  \frac{\sqrt{(n-1)} \times \sum_{i=1}^{n}\left ( x_{i} y_{i} \right )}{\sqrt{\sum_{j=1}^{n} x_{j}^{2} \times \sum_{j=1}^{n} y_{j}^{2} - (\sum_{i=1}^{n}\left ( x_{i} y_{i} \right ))^2}} $$
